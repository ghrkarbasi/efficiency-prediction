```python
from sklearn.model_selection import train_test_split

train_set, test_set = train_test_split(data, test_size=0.2, random_state=42)

train_set.head(10) 
```

```python
train_df = train_set.copy()

train_df_label = train_df["bcc score"].copy()  #==========df_label=========
train_df_num = train_df.drop("bcc score",axis=1)

train_df.info()

#*********************************************************************************
print("\n\n")
test_df = test_set.copy()

test_df_label = test_df["bcc score"].copy()  #==========df_label=========
test_df_num = test_df.drop("bcc score",axis=1)

test_df.info()
```

```python
#===========================feature scaling=================
#standardization z=(x-mu)/sigma      normalisation [0,1] x_norm=(x-x_min)/(x_max-x_min)

from sklearn.preprocessing import StandardScaler

feature_scal = StandardScaler()

train_df_num_scaled = pd.DataFrame(
                        feature_scal.fit_transform(train_df_num.values),
                        columns = train_df_num.columns)

print(train_df_num_scaled.head())
#**********************************************************************

test_df_num_scaled = pd.DataFrame(
                        feature_scal.fit_transform(test_df_num.values),
                        columns = test_df_num.columns)

print("\n\n",test_df_num_scaled.head())
```

```python
#============================================the last output file is:::: df_num_impute_tr ===========        
import numpy as np

new_train_df_standard = np.c_[train_df_num_scaled, train_df_label]

new_train_df_standard = pd.DataFrame(
                        new_train_df_standard)
columns = list(train_df_num_scaled.columns)
columns.append("bcc score")

new_train_df_standard.columns = columns

new_train_df_standard.head(10)
```

```python
features = ["bcc score","net profit","owners equity","operational income", "total operational expences"]
scatter_matrix(new_train_df_standard[features],  figsize=(15, 10))
plt.show()
```

```python
new_train_df_standard.plot(kind='scatter',x='net profit',y='bcc score',
          figsize=(10,7), alpha=0.2)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(train_df_num_scaled, train_df_label)
sampel_data_prepared = train_df_num_scaled.iloc[:4]
print("Predictions:", lin_reg.predict(sampel_data_prepared))
sampel_labels = train_df_label.iloc[:4]
print("Labels:\t", list(sampel_labels))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error #RMSE => Root mean squred error
import numpy as np

lin_reg = LinearRegression()
lin_reg.fit(train_df_num_scaled, train_df_label)
effi_predictions = lin_reg.predict(train_df_num_scaled)
lin_mse = mean_squared_error(train_df_label, effi_predictions)
lin_rmse = np.sqrt(lin_mse)
lin_rmse
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor()
tree_reg.fit(train_df_num_scaled, train_df_label)
effi_predictions = tree_reg.predict(train_df_num_scaled)
tree_mse = mean_squared_error(train_df_label, effi_predictions)
tree_rmse = np.sqrt(tree_mse)
print(tree_rmse)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(tree_reg, train_df_num_scaled, train_df_label, scoring = "neg_mean_squared_error", cv = 10)
trees_rmse_scores = np.sqrt(-scores)
def display_scores(scores, model_name):
    print("================", model_name, '================')
    print("scores:", scores)
    print("mean:",scores.mean())
    print("standard deviation:", scores.std())
    print("==========================================================")
display_scores(trees_rmse_scores, "decision tree regression")
```

```python
linear_scores = cross_val_score(lin_reg, train_df_num_scaled, train_df_label, scoring = "neg_mean_squared_error", cv = 10)
linear_rmse_scores = np.sqrt(-linear_scores)
display_scores(linear_rmse_scores, "linear regression")
```

```python
from sklearn.ensemble import RandomForestRegressor

forest_reg = RandomForestRegressor()
forest_reg.fit(train_df_num_scaled, train_df_label.values.ravel())
forest_scores = cross_val_score(forest_reg, train_df_num_scaled, train_df_label.values.ravel(), scoring = "neg_mean_squared_error", cv = 10)
forest_rmse_scores = np.sqrt(-forest_scores)
display_scores(forest_rmse_scores, "Random Forest Regressor")
```

```python
from catboost import CatBoostRegressor

catboost_reg = CatBoostRegressor(
                  #verbose=10, n_estimators=100
                                )
#catboost_reg.fit(train_df_num_scaled, train_df_label)
catboost_scores = cross_val_score(catboost_reg, train_df_num_scaled, train_df_label, scoring = "neg_mean_squared_error", cv = 10)
catboost_rmse_scores = np.sqrt(-catboost_scores)
display_scores(catboost_rmse_scores, "CatBoost Regressor")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

param_grid = [{'n_estimators': [3, 4, 6, 10, 30], 'max_features': [2, 6, 8, 15]}]
forest_reg = RandomForestRegressor()

grid_search = GridSearchCV(forest_reg, param_grid, cv = 5, scoring = 'neg_mean_squared_error')
grid_search.fit(train_df_num_scaled, train_df_label.values.ravel())

print("best parameters:", grid_search.best_params_)
print("best estimator:", grid_search.best_estimator_)
```

```python
results = grid_search.cv_results_

for mean_score, params in zip(results["mean_test_score"], results["params"]):
    print(np.sqrt(-mean_score), params)
```

```python
#===========================test final model on the test set===========
final_model = grid_search.best_estimator_

final_predictions = final_model.predict(test_df_num_scaled)
final_mse = mean_squared_error(test_df_label, final_predictions)
final_rmse = np.sqrt(final_mse)
final_rmse
```
